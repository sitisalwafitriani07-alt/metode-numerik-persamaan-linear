```python

import math  # Mengimpor modul math untuk perhitungan matematika.

# Mendefinisikan fungsi persamaan nonlinear f(t).
def f(t):
    # Menghitung nilai fungsi f(t) sesuai persamaan pada soal.
    return 800 * math.exp(0.3 * t) - 2000 - 400 * t

# Menentukan interval awal pencarian akar.
a = 5.0  # Menentukan batas bawah interval.
b = 6.0  # Menentukan batas atas interval.

# Menentukan batas ketelitian perhitungan.
toleransi = 0.0001

# Menginisialisasi jumlah iterasi dari nol.
iterasi = 0

# Memeriksa apakah interval awal memenuhi syarat metode Biseksi.
# Hasil perkalian harus negatif agar terdapat perubahan tanda.
if f(a) * f(b) >= 0:
    # Menampilkan pesan jika interval tidak memenuhi syarat.
    print("Interval tidak memenuhi syarat metode Biseksi.")

else:
    # Menampilkan judul tabel hasil perhitungan.
    print("TABEL ITERASI METODE BISEKSI")

    # Menampilkan garis pembatas tabel sepanjang 65 karakter.
    print("-" * 65)

    # Menampilkan nama kolom tabel.
    # <8 berarti rata kiri dengan lebar 8 karakter.
    # >10 berarti rata kanan dengan lebar 10 karakter.
    print(f"{'Iterasi':<8}{'a':>10}{'b':>10}"
          f"{'c':>12}{'f(c)':>15}")

    # Menampilkan garis pembatas di bawah judul kolom.
    print("-" * 65)

    # Mengulang perhitungan selama lebar interval lebih besar
    # daripada nilai toleransi yang telah ditentukan.
    while (b - a) > toleransi:

        # Menambah penghitung iterasi sebanyak satu.
        iterasi += 1

        # Menghitung titik tengah interval menggunakan rumus Biseksi.
        c = (a + b) / 2

        # Menampilkan hasil iterasi, batas bawah, batas atas,
        # titik tengah, dan nilai fungsi pada titik tengah.
        # Format .4f menampilkan empat angka di belakang koma.
        print(f"{iterasi:<8}{a:>10.4f}{b:>10.4f}"
              f"{c:>12.4f}{f(c):>15.4f}")

        # Memeriksa perubahan tanda antara f(a) dan f(c).
        if f(a) * f(c) < 0:

            # Jika hasil perkalian negatif, akar berada
            # di antara a dan c, sehingga batas atas diubah.
            b = c

        else:
            # Jika hasil perkalian tidak negatif,
            # batas bawah diperbarui menjadi titik tengah.
            a = c

    # Menghitung perkiraan akar dari interval terakhir.
    akar = (a + b) / 2

    # Menghitung biaya cloud pada waktu perkiraan titik impas.
    biaya_cloud = 800 * math.exp(0.3 * akar)

    # Menghitung biaya server sendiri pada waktu yang sama.
    biaya_server = 2000 + 400 * akar

    # Menampilkan garis pemisah sebelum hasil akhir.
    print("-" * 65)

    # Menampilkan jumlah iterasi yang telah dilakukan.
    print(f"Jumlah iterasi : {iterasi}")

    # Menampilkan perkiraan waktu titik impas dalam bulan.
    print(f"Perkiraan akar : {akar:.4f} bulan")

    # Menampilkan perkiraan biaya layanan cloud.
    print(f"Biaya cloud    : ${biaya_cloud:.4f}")

    # Menampilkan perkiraan biaya server sendiri.
    print(f"Biaya server   : ${biaya_server:.4f}")
```
